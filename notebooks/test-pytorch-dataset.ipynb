```python
from typing import List,Tuple
import polars as pl
from tqdm import tqdm
```

```python
ldf = pl.scan_csv(files_handler.csv_file_path)

def collect(df:pl.LazyFrame, offset:int) -> List[Tuple[str,int]]:
    collected = ldf.slice(offset=offset, length=10).select(["file", "index"]).collect().to_numpy().tolist()
    print(collected)
    return collected
collect(ldf,0)
collect(ldf,1)
```

```python
for p,i in collect(ldf,0):
    print(p,i)
```
